# Presentaciones ejecutivas

Presentación ejecutiva para el directorio de una cadena de supermercados con datos de ventas individuales (producto, trimestre, venta, satisfacción, cliente_id y margen) que sirve como caso único para aplicar estructura narrativa, mensaje central, honestidad visual e incertidumbre.


## Mensaje central único

La comunicación concisa y directa evita la información innecesaria y concentra el mensaje principal en el titular de los mensajes, las diapositivas o las secciones de un informe. Colocar la idea central en el encabezado permite que quien lee procese la información con mayor rapidez y decida si el contenido es relevante y si conviene profundizar en él. Así, el titular deja de ser una etiqueta decorativa y pasa a cumplir la función de comunicar lo esencial antes que cualquier detalle secundario.

**Ejemplo:** En la presentación ejecutiva para el directorio de la cadena de supermercados, el titular de la diapositiva de ventas enuncia directamente que el margen cayó en el trimestre, en lugar de rotularla solo como "Ventas"; así el directorio capta el mensaje central de inmediato y decide si revisa el detalle por producto.

```python
import matplotlib.pyplot as plt

# Datos ilustrativos del margen por trimestre
trimestres = ['T1', 'T2', 'T3', 'T4']
margen = [18.5, 17.2, 15.8, 14.1]

# Creamos la figura y el eje
fig, ax = plt.subplots(figsize=(8, 5))

# Graficamos la evolución del margen con barras
ax.bar(trimestres, margen, color='#c0392b')

# Titular que comunica el mensaje central directamente
ax.set_title('El margen cayó en el trimestre', fontsize=15, fontweight='bold')

# Etiquetas de los ejes
ax.set_xlabel('Trimestre')
ax.set_ylabel('Margen (%)')

# Mostramos el gráfico
plt.show()
```

**Errores frecuentes:**
- Incluir información innecesaria en la comunicación
- No colocar el mensaje principal en el titular de mensajes, diapositivas o secciones del informe

## Estructura piramidal ascendente

La estructura piramidal ascendente consiste en abrir la comunicación con la llamada a la acción, es decir, con aquello que se espera del público. Esta elección tiene una ventaja doble: por un lado, la audiencia comprende de inmediato cuál es su papel y desde qué perspectiva debe interpretar el resto del contenido; por otro, entiende por qué conviene seguir prestando atención. Al definir primero la acción esperada, se orienta la lectura o escucha de todo lo que sigue y se evita que el receptor tenga que inferir el propósito. La claridad sobre el rol y el prisma de análisis funciona como marco para el resto de la presentación o comunicación, de modo que cada parte posterior se interpreta en función de esa expectativa inicial.

**Ejemplo:** Ante el directorio de la cadena de supermercados, se abre la presentación con la acción esperada: aprobar el presupuesto para la campaña del próximo trimestre. Recién después se muestran las ventas por producto, la satisfacción de los clientes y el margen, de modo que los directivos interpretan cada dato desde ese pedido concreto y saben por qué deben seguir escuchando.

```python
import matplotlib.pyplot as plt

# Datos ilustrativos de la presentación al directorio
productos = ['Lácteos', 'Bebidas', 'Almacén', 'Frescos']
ventas = [120, 95, 80, 110]
satisfaccion = [85, 78, 82, 90]
margen = [22, 18, 25, 20]

# La acción esperada se muestra primero, como marco de toda la presentación
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Panel izquierdo: la llamada a la acción que abre la comunicación
ax1.text(0.5, 0.5, 'ACCIÓN ESPERADA:\nAprobar el presupuesto\npara la campaña del\npróximo trimestre',
         ha='center', va='center', fontsize=14, fontweight='bold',
         bbox=dict(boxstyle='round,pad=0.6', facecolor='#cfe2f3', edgecolor='#1f4e79'))
ax1.set_xlim(0, 1)
ax1.set_ylim(0, 1)
ax1.axis('off')
ax1.set_title('1. Apertura: llamada a la acción')

# Panel derecho: los datos que se interpretan desde ese pedido concreto
x = range(len(productos))
ancho = 0.35
ax2.bar([i - ancho/2 for i in x], ventas, width=ancho, label='Ventas (miles)', color='#4c72b0')
ax2.bar([i + ancho/2 for i in x], margen, width=ancho, label='Margen (%)', color='#dd8452')
ax2.set_xticks(list(x))
ax2.set_xticklabels(productos)
ax2.set_ylabel('Valor')
ax2.set_title('2. Datos de apoyo: ventas y margen por producto')
ax2.legend()

plt.tight_layout()
plt.show()
```

## Narrativa de recomendación

La narrativa de recomendación exige cuidar el orden y el estilo con que se presentan los datos. Una historia debe tener un planteamiento claro que funcione como argumento, un nudo con giros que sostengan la tensión y un desenlace que culmine en una llamada a la acción. El conflicto y la tensión son recursos para captar y mantener la atención de la audiencia. El orden de los elementos y el estilo de la narración condicionan la claridad del mensaje, por lo que conviene considerar ambas dimensiones al construir la historia. La repetición, bien empleada, favorece que las historias se interioricen. Tácticas como las lógicas horizontal y vertical o la representación visual del guion inversa, junto con la búsqueda de una nueva perspectiva, ayudan a que la historia se transmita con claridad en la presentación.

**Ejemplo:** Al preparar la presentación ejecutiva para el directorio de la cadena de supermercados, se ordenan las ventas por trimestre para abrir con el argumento del crecimiento, se intercalan los giros de satisfacción y margen para generar tensión y se cierra con una llamada a la acción sobre los productos de bajo rendimiento.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos de la presentación ejecutiva por trimestre
datos = {
    'trimestre': ['T1', 'T2', 'T3', 'T4'],
    'ventas': [100, 120, 135, 150],
    'satisfaccion': [70, 75, 68, 80],
    'margen': [25, 24, 22, 21]
}
df = pd.DataFrame(datos)

# Planteamiento: abrir con el argumento del crecimiento de ventas
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(df['trimestre'], df['ventas'], marker='o', color='tab:blue', label='Ventas (índice)')

# Nudo: intercalar los giros de satisfacción y margen para generar tensión
ax.plot(df['trimestre'], df['satisfaccion'], marker='s', color='tab:orange', label='Satisfacción')
ax.plot(df['trimestre'], df['margen'], marker='^', color='tab:green', label='Margen')

# Desenlace: señalar la llamada a la acción sobre los productos de bajo rendimiento
ax.annotate('Llamada a la acción:\nrevisar productos de bajo rendimiento',
            xy=('T4', 21), xytext=('T2', 40),
            arrowprops=dict(arrowstyle='->', color='tab:red'),
            color='tab:red')

# Estilo y orden de la narrativa
ax.set_title('Narrativa de recomendación: ventas, satisfacción y margen por trimestre')
ax.set_xlabel('Trimestre')
ax.set_ylabel('Valor')
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```

## Gestión del tiempo y preguntas

Al finalizar una presentación, conviene evaluar si se compartirán las diapositivas con la audiencia y de qué manera. Además, tanto al exponer como al criticar una presentación, es útil plantear una serie de preguntas para valorar su calidad. Entre ellas, se considera si el orador eligió una historia interesante para contar con los datos, en relación con el tema y la audiencia. Estas preguntas orientan la reflexión sobre la efectividad de la presentación y ayudan a identificar aspectos de mejora.

**Ejemplo:** En la presentación ejecutiva para el directorio de la cadena de supermercados, después de exponer, se plantea si se compartirán las diapositivas con los directores y se evalúa si la historia contada con los datos de ventas, satisfacción y margen resultó interesante y adecuada para ese auditorio.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos de la presentación ejecutiva para el directorio
datos = {
    'Indicador': ['Ventas', 'Satisfacción', 'Margen'],
    'Valor': [85, 72, 64]
}
df = pd.DataFrame(datos)

# Gráfico de barras con los indicadores presentados al directorio
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(df['Indicador'], df['Valor'], color=['#4C72B0', '#55A868', '#C44E52'])
ax.set_title('Indicadores presentados al directorio')
ax.set_xlabel('Indicador')
ax.set_ylabel('Valor (%)')
ax.set_ylim(0, 100)

# Preguntas de evaluación que se plantean al finalizar la presentación
preguntas = [
    '¿Se compartirán las diapositivas con los directores?',
    '¿La historia contada con los datos resultó interesante?',
    '¿Fue adecuada para este auditorio?'
]

# Mostrar las preguntas debajo del gráfico
texto_preguntas = '\n'.join(['• ' + p for p in preguntas])
fig.text(0.5, -0.05, texto_preguntas, ha='center', va='top', fontsize=10)

plt.tight_layout()
plt.show()
```

## Ética de la evidencia selectiva

Representar los datos de forma inexacta para reforzar un argumento constituye un engaño y no es correcto. Más allá de las consideraciones morales, se trata de un terreno arriesgado: si un solo experto entre el público advierte la manipulación, por ejemplo que el eje y de un gráfico de barras empieza en un valor distinto de cero, todo el argumento queda en evidencia y también la credibilidad de quien lo presenta. La ética de la evidencia selectiva exige, por tanto, no alterar la escala ni manipular los datos para favorecer una conclusión, porque el costo reputacional puede ser alto.

**Ejemplo:** Al mostrar la evolución trimestral de las ventas, se ajusta el eje y para que empiece en un valor alto y las diferencias parezcan mayores; si un miembro del directorio detecta que la escala no parte de cero, la conclusión sobre el crecimiento pierde fuerza y la credibilidad del expositor queda cuestionada.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos: evolución trimestral de las ventas
trimestres = ['T1', 'T2', 'T3', 'T4']
ventas = [100, 104, 107, 110]
df = pd.DataFrame({'trimestre': trimestres, 'ventas': ventas})

# Creamos dos gráficos para comparar la versión manipulada con la correcta
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Gráfico engañoso: el eje y empieza en 95, exagerando las diferencias
ax1.bar(df['trimestre'], df['ventas'], color='tomato')
ax1.set_ylim(95, 112)
ax1.set_title('Gráfico engañoso: eje y no parte de cero')
ax1.set_xlabel('Trimestre')
ax1.set_ylabel('Ventas (miles)')

# Gráfico correcto: el eje y empieza en cero, mostrando la escala real
ax2.bar(df['trimestre'], df['ventas'], color='steelblue')
ax2.set_ylim(0, 120)
ax2.set_title('Gráfico correcto: eje y parte de cero')
ax2.set_xlabel('Trimestre')
ax2.set_ylabel('Ventas (miles)')

plt.tight_layout()
plt.show()
```

**Errores frecuentes:**
- Cambiar la escala de un gráfico de barras para reforzar el argumento
- Manipular los datos para representarlos de forma inexacta

### Solución


```python
import matplotlib.pyplot as plt

# Datos de ventas trimestrales (en millones)
trimestres = ['Q1', 'Q2', 'Q3', 'Q4']
ventas = [120, 150, 130, 180]

# Ordenar los datos de menor a mayor para una narrativa de progresión
pares = sorted(zip(ventas, trimestres))
ventas_ordenadas = [v for v, t in pares]
trimestres_ordenados = [t for v, t in pares]

plt.bar(trimestres_ordenados, ventas_ordenadas, color='skyblue')
plt.title('Las ventas crecieron un 50% en el año')  # Titular con el mensaje central
plt.ylim(0, 200)  # Eje y comienza en cero para no distorsionar
plt.xlabel('Trimestre')
plt.ylabel('Ventas (millones)')
plt.show()
```
