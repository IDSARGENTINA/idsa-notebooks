# Honestidad visual: ejes y escalas engañosas

Presentación ejecutiva para el directorio de una cadena de supermercados con datos de ventas individuales (producto, trimestre, venta, satisfacción, cliente_id y margen) que sirve como caso único para aplicar estructura narrativa, mensaje central, honestidad visual e incertidumbre.


## Eje truncado y línea base

Si bien en algunos gráficos de series temporales se acepta una línea base distinta de cero, en los gráficos de barras verticales se debe utilizar un eje y con punto cero. Un eje truncado puede exagerar la diferencia entre valores: el gráfico de la izquierda de un ejemplo exagera la diferencia real del 5,6% al iniciar el eje y en 101. En el ejemplo, la diferencia de altura de las barras entre 2011 y 2012 muestra un efecto de tamaño del 122%, lo que da un factor de mentira (lie factor) de 21,78. Para preservar la integridad de los datos, el porcentaje mostrado en el gráfico debe corresponder a los datos. Por lo tanto, en las barras verticales conviene usar siempre un eje y que parta de cero, mientras que en algunos gráficos de series temporales una línea base distinta de cero resulta aceptable.

**Ejemplo:** Datos ilustrativos: una empresa ficticia presenta ventas trimestrales de 100, 102 y 105 unidades. Si se usa un gráfico de barras verticales con eje desde 100, la barra de 105 parece cinco veces más alta que la de 102, aunque la diferencia real es de solo 3 unidades. Para evitar esta distorsión, se debe iniciar el eje en cero, de modo que las alturas reflejen proporcionalmente los valores reales.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos: ventas trimestrales de una empresa ficticia
trimestres = ['T1', 'T2', 'T3']
ventas = [100, 102, 105]

df = pd.DataFrame({'Trimestre': trimestres, 'Ventas': ventas})

# Creamos una figura con dos subplots lado a lado
fig, (ax_izq, ax_der) = plt.subplots(1, 2, figsize=(12, 5))

# --- Gráfico de la IZQUIERDA: eje truncado (empieza en 100) ---
# Esto exagera la diferencia entre valores
ax_izq.bar(df['Trimestre'], df['Ventas'], color='indianred')
ax_izq.set_ylim(100, 106)  # eje y truncado, NO parte de cero
ax_izq.set_title('Eje truncado (desde 100)\nExagera la diferencia')
ax_izq.set_ylabel('Ventas (unidades)')
ax_izq.set_xlabel('Trimestre')

# Anotamos los valores reales sobre cada barra
for i, v in enumerate(df['Ventas']):
    ax_izq.text(i, v + 0.1, str(v), ha='center', va='bottom')

# --- Gráfico de la DERECHA: eje desde cero (correcto) ---
# Las alturas reflejan proporcionalmente los valores reales
ax_der.bar(df['Trimestre'], df['Ventas'], color='steelblue')
ax_der.set_ylim(0, 115)  # eje y parte de cero
ax_der.set_title('Eje desde cero (correcto)\nProporciones reales')
ax_der.set_ylabel('Ventas (unidades)')
ax_der.set_xlabel('Trimestre')

# Anotamos los valores reales sobre cada barra
for i, v in enumerate(df['Ventas']):
    ax_der.text(i, v + 1, str(v), ha='center', va='bottom')

plt.tight_layout()
plt.savefig('eje_truncado_vs_cero.png', dpi=100)
plt.close()
```

**Errores frecuentes:**
- Iniciar el eje y en un valor distinto de cero en gráficos de barras verticales, lo que exagera las diferencias entre valores.
- Mostrar en el gráfico un porcentaje de diferencia que no coincide con la variación real de los datos, afectando la integridad de la información.

## Escala logarítmica mal usada

En una escala logarítmica la multiplicación se ve como una suma en una escala lineal, por lo que estas escalas son la opción natural para datos obtenidos por multiplicación o división. En particular, las razones o ratios deben mostrarse generalmente en escala logarítmica. Lo ilustra con el ejemplo de los habitantes de los condados de Texas divididos por la mediana: al visualizar esos ratios en escala logarítmica, se observa claramente que las poblaciones se distribuyen de forma simétrica alrededor de la mediana, y que los condados más poblados tienen más de 100 veces la población de la mediana, mientras que los menos poblados tienen más de 100 veces menos. En contraste, para los mismos datos, una escala lineal oculta las diferencias entre un condado con población mediana y uno con una población mucho menor que la mediana. Por tanto, en el caso de los ratios, la escala logarítmica permite apreciar claramente las diferencias y la simetría de la distribución, mientras que la escala lineal las oculta.

**Ejemplo:** Datos ilustrativos (ficticios): supongamos que en una cadena de supermercados se calcula el ratio de ventas de cada producto respecto a la mediana de ventas de todos los productos. Si los ratios son 0.01, 0.1, 1, 10 y 100, al graficarlos en escala logarítmica se aprecia que están simétricamente distribuidos alrededor de 1 (la mediana), y que los productos con ratios extremos se alejan en órdenes de magnitud. En cambio, en escala lineal, los ratios 0.01, 0.1 y 1 quedarían amontonados cerca del cero, ocultando las diferencias entre ellos, mientras que el ratio 100 dominaría la gráfica.

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos ilustrativos: ratios de ventas de cada producto respecto a la mediana
ratios = [0.01, 0.1, 1, 10, 100]
etiquetas = ['Producto A', 'Producto B', 'Producto C', 'Producto D', 'Producto E']

# Creamos una figura con dos subplots: escala lineal y escala logarítmica
fig, (ax_lineal, ax_log) = plt.subplots(1, 2, figsize=(12, 5))

# --- Panel izquierdo: escala lineal (oculta las diferencias) ---
ax_lineal.bar(etiquetas, ratios, color='steelblue')
ax_lineal.set_title('Escala lineal\n(oculta las diferencias)')
ax_lineal.set_ylabel('Ratio respecto a la mediana')
ax_lineal.set_xlabel('Producto')
ax_lineal.axhline(1, color='red', linestyle='--', label='Mediana (ratio = 1)')
ax_lineal.legend()

# --- Panel derecho: escala logarítmica (muestra la simetría) ---
ax_log.bar(etiquetas, ratios, color='darkorange')
ax_log.set_yscale('log')
ax_log.set_title('Escala logarítmica\n(muestra simetría alrededor de 1)')
ax_log.set_ylabel('Ratio respecto a la mediana (escala log)')
ax_log.set_xlabel('Producto')
ax_log.axhline(1, color='red', linestyle='--', label='Mediana (ratio = 1)')
ax_log.legend()

# Ajustamos el layout y guardamos la figura
plt.tight_layout()
plt.savefig('escala_logaritmica_ratios.png', dpi=100)
plt.close()
```

**Errores frecuentes:**
- Usar una escala lineal para representar ratios, lo que oscurece las diferencias entre valores pequeños y la simetría alrededor de la mediana.
- Emplear una escala logarítmica con datos que no provienen de multiplicaciones o divisiones, lo que puede distorsionar la interpretación.

## Ejes duales y correlación falsa

En las preguntas causales normalmente podemos argumentar en ambos sentidos: que X causa Y, o que una tercera variable Z causa tanto X como Y, de modo que la correlación entre X e Y es espuria. Añade que, por esta razón, conocer el mecanismo de asignación conduce a una respuesta causal mucho más convincente. Es decir, la mera correlación entre dos variables no basta para decidir la dirección causal, porque siempre es posible postular una variable de confusión Z que explique ambas. Por eso, en inferencia causal aplicada se requiere pensar seriamente sobre el mecanismo que genera los datos, a diferencia del aprendizaje automático aplicado que suele consistir en pulsar botones en el orden correcto.

**Ejemplo:** Supongamos (datos ilustrativos) que en una cadena de supermercados se observa que los productos con mayor venta trimestral también tienen mayor satisfacción del cliente. Podríamos argumentar que vender más causa mayor satisfacción, o que una tercera variable, como la calidad del producto, causa tanto mayores ventas como mayor satisfacción, haciendo espuria la correlación. Sin conocer el mecanismo de asignación, ambos argumentos son plausibles.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Datos ilustrativos del ejemplo: productos de supermercado
# Cada fila: producto, ventas trimestrales (unidades), satisfaccion del cliente (0-100), calidad del producto (0-100)
datos = pd.DataFrame({
    'producto': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'ventas': [120, 200, 150, 300, 80, 250, 180, 220],
    'satisfaccion': [60, 75, 65, 90, 50, 85, 70, 80],
    'calidad': [55, 78, 62, 92, 48, 88, 68, 82]
})

# Creamos una figura con dos ejes duales (subplots) para mostrar dos lecturas causales
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# --- Panel izquierdo: correlacion directa ventas vs satisfaccion ---
ax1.scatter(datos['ventas'], datos['satisfaccion'], color='tab:blue', s=80)
for i, row in datos.iterrows():
    ax1.annotate(row['producto'], (row['ventas'], row['satisfaccion']),
                 textcoords='offset points', xytext=(5, 5), fontsize=9)
ax1.set_xlabel('Ventas trimestrales (unidades)')
ax1.set_ylabel('Satisfaccion del cliente')
ax1.set_title('Correlacion observada: ventas vs satisfaccion')
ax1.grid(True, alpha=0.3)

# --- Panel derecho: la variable de confusion (calidad) explica ambas ---
# Eje dual: ventas en eje izquierdo, satisfaccion en eje derecho, ambas contra calidad
ax2.scatter(datos['calidad'], datos['ventas'], color='tab:green', s=80, label='Ventas')
ax2.set_xlabel('Calidad del producto (variable de confusion Z)')
ax2.set_ylabel('Ventas trimestrales', color='tab:green')
ax2.tick_params(axis='y', labelcolor='tab:green')
ax2.grid(True, alpha=0.3)

# Segundo eje Y para la satisfaccion
ax2b = ax2.twinx()
ax2b.scatter(datos['calidad'], datos['satisfaccion'], color='tab:red', s=80, marker='s', label='Satisfaccion')
ax2b.set_ylabel('Satisfaccion del cliente', color='tab:red')
ax2b.tick_params(axis='y', labelcolor='tab:red')

ax2.set_title('Variable de confusion Z (calidad) causa ambas')

# Leyenda combinada
lines1, labels1 = ax2.get_legend_handles_labels()
lines2, labels2 = ax2b.get_legend_handles_labels()
ax2.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.tight_layout()
plt.savefig('ejes_duales_correlacion_espuria.png', dpi=100)
print('Figura guardada: ejes_duales_correlacion_espuria.png')
```

## Inversión y omisión de ejes

Indica que Plotly permite invertir la dirección de un eje configurando la propiedad autorange con el valor "reversed". Esto significa que, en lugar de que el eje vaya de menor a mayor (o de abajo hacia arriba en el eje y), el rango automático se calcula en sentido contrario. Muestra un ejemplo con un gráfico de dispersión facetado por especie, donde se aplica update_yaxes(autorange="reversed") para invertir el eje y. Así, el concepto se centra en cómo modificar la orientación de un eje sin cambiar los datos, simplemente ajustando la lógica de cálculo automático del rango. Es una herramienta útil para adaptar la presentación visual según el propósito del gráfico.

**Ejemplo:** Supongamos un gráfico de dispersión facetado por categoría de producto (datos ilustrativos) que muestra en el eje y el margen de ganancia. Por defecto, el eje y va de menor a mayor; al aplicar autorange="reversed", los valores mayores aparecen abajo y los menores arriba, invirtiendo la lectura visual. Esto puede usarse para enfatizar ciertos patrones, aunque los datos no cambian.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos: dispersión facetada por categoría de producto
datos = pd.DataFrame({
    'categoria': ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C'],
    'x': [1, 2, 3, 1, 2, 3, 1, 2, 3],
    'margen': [10, 25, 40, 15, 30, 50, 5, 20, 35]
})

categorias = ['A', 'B', 'C']

# Creamos una figura con dos filas: arriba eje normal, abajo eje invertido
fig, axes = plt.subplots(2, 3, figsize=(12, 7), sharex=True)

for i, cat in enumerate(categorias):
    sub = datos[datos['categoria'] == cat]
    # Fila superior: eje y normal (menor a mayor)
    axes[0, i].scatter(sub['x'], sub['margen'], color='tab:blue')
    axes[0, i].set_title(f'Categoría {cat} - eje normal')
    axes[0, i].set_ylabel('Margen de ganancia')
    # Fila inferior: eje y invertido (autorange='reversed')
    axes[1, i].scatter(sub['x'], sub['margen'], color='tab:orange')
    axes[1, i].set_title(f'Categoría {cat} - eje invertido')
    axes[1, i].set_ylabel('Margen de ganancia')
    axes[1, i].invert_yaxis()  # equivale a autorange='reversed'
    axes[1, i].set_xlabel('x')

plt.tight_layout()
plt.savefig('ejes_invertidos.png')
```
