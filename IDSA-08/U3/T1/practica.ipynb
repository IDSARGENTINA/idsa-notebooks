# Estructura narrativa

Presentación ejecutiva para el directorio de una cadena de supermercados con datos de ventas individuales (producto, trimestre, venta, satisfacción, cliente_id y margen) que sirve como caso único para aplicar estructura narrativa, mensaje central, honestidad visual e incertidumbre.


## Arco narrativo en tres actos

El arco del personaje es un componente del segundo acto del relato, donde se narra la mayor parte de la historia. En este acto se describe el intento del protagonista de resolver el problema creado por el primer punto de inflexión. A menudo, al personaje principal le faltan recursos para gestionar el conflicto, lo que lo lleva a enfrentar situaciones cada vez peores. Durante este proceso, el protagonista experimenta transformaciones importantes en su vida como resultado de lo que ocurre. Así, el arco del personaje refleja el cambio interno que surge de las dificultades enfrentadas en el segundo acto.

**Ejemplo:** En una presentación ficticia para el directorio de una cadena de supermercados, se podría ilustrar el arco del personaje con un gerente que, tras recibir datos de ventas trimestrales (producto, venta, satisfacción, cliente_id y margen), intenta resolver la caída de ventas con recursos limitados. A medida que enfrenta situaciones cada vez peores, como la insatisfacción de los clientes, experimenta una transformación en su enfoque de gestión. (Datos ilustrativos).

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos de ventas trimestrales de la cadena de supermercados
datos = {
    'trimestre': ['T1', 'T2', 'T3', 'T4'],
    'venta': [120000, 98000, 82000, 75000],
    'satisfaccion': [85, 72, 60, 55],
    'margen': [0.22, 0.19, 0.15, 0.12],
    'cliente_id': [1001, 1002, 1003, 1004]
}
df = pd.DataFrame(datos)

# Creamos la figura con dos subplots: ventas y satisfaccion
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Grafico de ventas trimestrales (caida progresiva)
ax1.plot(df['trimestre'], df['venta'], marker='o', color='tab:blue', linewidth=2)
ax1.set_title('Ventas trimestrales (caida)')
ax1.set_xlabel('Trimestre')
ax1.set_ylabel('Venta (unidades)')
ax1.grid(True, alpha=0.3)

# Grafico de satisfaccion del cliente (deterioro)
ax2.plot(df['trimestre'], df['satisfaccion'], marker='s', color='tab:red', linewidth=2)
ax2.set_title('Satisfaccion del cliente')
ax2.set_xlabel('Trimestre')
ax2.set_ylabel('Satisfaccion (%)')
ax2.grid(True, alpha=0.3)

# Anotamos el arco del personaje: el gerente enfrenta situaciones cada vez peores
ax1.annotate('Primer punto de inflexion\n(caida de ventas)', xy=('T2', 98000),
             xytext=('T1', 110000), arrowprops=dict(arrowstyle='->', color='gray'))
ax2.annotate('Transformacion del enfoque\nde gestion', xy=('T4', 55),
             xytext=('T2', 65), arrowprops=dict(arrowstyle='->', color='gray'))

plt.tight_layout()
plt.savefig('arco_personaje.png')
print('Figura guardada como arco_personaje.png')
```

## Conflicto y tensión dramática

El conflicto y la tensión dramática son elementos esenciales en una historia, según se afirma en. Las historias sin conflicto, donde todo es perfecto, no resultan interesantes ni inspiran acción. Por eso, se debe pensar en el conflicto y la tensión como herramientas de storytelling que ayudan a implicar al público. Además, se sugiere enmarcar la historia en el contexto del problema del público, para que se sientan partícipes de la solución. Nancy Duarte llama a esta tensión 'el conflicto entre lo que es y lo que podría ser'. Así, el conflicto y la tensión son cruciales para captar la atención y motivar a la audiencia.

**Ejemplo:** En una presentación ejecutiva para el directorio de una cadena de supermercados (datos ilustrativos), se podría mostrar la tensión entre la situación actual de ventas (lo que es) y la meta de crecimiento (lo que podría ser), usando datos ficticios de ventas trimestrales para ilustrar el conflicto y generar interés en la solución propuesta.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos: ventas trimestrales (en millones) de una cadena de supermercados
trimestres = ['Q1', 'Q2', 'Q3', 'Q4']
ventas_actuales = [120, 125, 118, 130]   # lo que es
meta_crecimiento = [150, 155, 160, 165]  # lo que podría ser

df = pd.DataFrame({
    'Trimestre': trimestres,
    'Ventas actuales': ventas_actuales,
    'Meta de crecimiento': meta_crecimiento
})

# Crear figura 2D
fig, ax = plt.subplots(figsize=(8, 5))

# Graficar las dos series para mostrar la tensión entre lo que es y lo que podría ser
ax.plot(df['Trimestre'], df['Ventas actuales'], marker='o', label='Lo que es (ventas actuales)', color='tab:red')
ax.plot(df['Trimestre'], df['Meta de crecimiento'], marker='s', label='Lo que podría ser (meta)', color='tab:green')

# Rellenar el área entre ambas líneas para visualizar la brecha (tensión dramática)
ax.fill_between(df['Trimestre'], df['Ventas actuales'], df['Meta de crecimiento'], color='gray', alpha=0.3, label='Brecha / tensión')

# Etiquetas y título
ax.set_xlabel('Trimestre')
ax.set_ylabel('Ventas (millones)')
ax.set_title('Conflicto y tensión dramática: situación actual vs. meta de crecimiento')
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)

# Guardar la figura (sin plt.show)
plt.tight_layout()
plt.savefig('tension_dramatica.png')
plt.close(fig)
```

## Narrador y punto de vista

Literal establece que, antes de convertir visualizaciones en historias, es necesario definir qué es una historia. Según, una historia es un conjunto de observaciones, hechos o eventos, verdaderos o inventados, presentados en un orden específico que genera una reacción emocional en la audiencia. Esa reacción se produce mediante la acumulación de tensión al inicio y una resolución hacia el final, lo que se conoce como arco narrativo. Destaca así que el orden en que se presentan los hechos es lo que permite construir ese arco y provocar la emoción deseada, y que toda buena historia tiene un arco claro e identificable. No menciona al narrador ni al punto de vista, por lo que este fragmento no permite desarrollar ese concepto.

**Ejemplo:** En una presentación ejecutiva (datos ilustrativos), el narrador podría comenzar mostrando una caída en las ventas trimestrales (tensión) y luego revelar un aumento en la satisfacción del cliente que anticipa una recuperación (resolución), guiando así la reacción emocional del directorio.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos de la presentación ejecutiva
trimestres = ['Q1', 'Q2', 'Q3', 'Q4']
ventas = [100, 85, 70, 60]  # caída en ventas (tensión)
satisfaccion = [70, 72, 80, 90]  # satisfacción del cliente que anticipa recuperación

fig, ax1 = plt.subplots(figsize=(8, 5))

# Eje 1: ventas (la tensión del arco narrativo)
ax1.plot(trimestres, ventas, marker='o', color='crimson', linewidth=2, label='Ventas')
ax1.set_xlabel('Trimestre')
ax1.set_ylabel('Ventas (índice)', color='crimson')
ax1.tick_params(axis='y', labelcolor='crimson')

# Eje 2: satisfacción del cliente (la resolución)
ax2 = ax1.twinx()
ax2.plot(trimestres, satisfaccion, marker='s', color='seagreen', linewidth=2, label='Satisfacción del cliente')
ax2.set_ylabel('Satisfacción del cliente', color='seagreen')
ax2.tick_params(axis='y', labelcolor='seagreen')

# Título que refleja el arco narrativo: tensión -> resolución
plt.title('Arco narrativo: caída de ventas y recuperación por satisfacción')

# Leyenda combinada de ambos ejes
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='center right')

plt.tight_layout()
plt.savefig('arco_narrativo.png')
```

## Trama frente a relato

Destaca la observación de Aristóteles de que un relato tiene planteamiento, nudo y desenlace, estructura que Vonnegut apreciaba. En el contexto del pasaje, esta idea se aplica a la comunicación con datos: el planteamiento es el primer acto donde se presenta la trama o argumento, estableciendo el contexto necesario para el público. Allí se definen los elementos esenciales de la historia (escenario, protagonista, situación no resuelta y resultado deseado) para que la audiencia comparta un marco común. Además, se debe involucrar a la audiencia despertando su interés y respondiendo a preguntas como por qué debería prestar atención o en qué le afecta. Así, la trama o argumento se convierte en el punto de partida del relato, siguiendo la estructura en tres actos inspirada en Aristóteles.

**Ejemplo:** En una presentación ejecutiva ficticia para el directorio de una cadena de supermercados, el planteamiento podría iniciar presentando la trama: 'Este trimestre, las ventas totales cayeron un 5% mientras la satisfacción del cliente se mantuvo estable'. Se establece el escenario (la cadena), el protagonista (los clientes y el equipo comercial), la situación no resuelta (la caída de ventas pese a la satisfacción) y el resultado deseado (recuperar el crecimiento). Con estos datos ilustrativos, se busca que la audiencia comprenda el contexto y se involucre desde el inicio.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del ejemplo: cadena de supermercados
# Ventas totales (índice base 100) y satisfacción del cliente (índice base 100)
datos = {
    'Trimestre': ['T1', 'T2', 'T3', 'T4'],
    'Ventas': [100, 98, 95, 95],          # caída del 5% respecto al inicio
    'Satisfaccion': [100, 100, 100, 100]  # satisfacción estable
}
df = pd.DataFrame(datos)

# --- PLANTEAMIENTO (primer acto) ---
# Se presenta la trama: escenario, protagonista, situación no resuelta y resultado deseado
fig, ax = plt.subplots(figsize=(9, 5))

# Graficamos la evolución de ventas y satisfacción
ax.plot(df['Trimestre'], df['Ventas'], marker='o', color='#c0392b',
        linewidth=2.5, label='Ventas totales (índice)')
ax.plot(df['Trimestre'], df['Satisfaccion'], marker='s', color='#2980b9',
        linewidth=2.5, linestyle='--', label='Satisfacción del cliente (índice)')

# Anotación de la situación no resuelta: caída de ventas con satisfacción estable
ax.annotate('Caída del 5% en ventas',
            xy=('T4', 95), xytext=('T2', 90),
            arrowprops=dict(arrowstyle='->', color='#c0392b'),
            color='#c0392b', fontsize=11, fontweight='bold')
ax.annotate('Satisfacción estable',
            xy=('T3', 100), xytext=('T1', 103),
            arrowprops=dict(arrowstyle='->', color='#2980b9'),
            color='#2980b9', fontsize=11, fontweight='bold')

# Título que enuncia la trama (el argumento del relato)
ax.set_title('Planteamiento: ventas caen 5% mientras la satisfacción se mantiene estable',
             fontsize=12, fontweight='bold')
ax.set_xlabel('Trimestre')
ax.set_ylabel('Índice (base 100)')
ax.set_ylim(85, 108)
ax.legend(loc='lower left')
ax.grid(True, alpha=0.3)

# Resultado deseado: recuperar el crecimiento (anotado como objetivo)
ax.text(0.98, 0.02, 'Resultado deseado: recuperar el crecimiento',
        transform=ax.transAxes, ha='right', va='bottom',
        fontsize=10, style='italic', color='#27ae60')

plt.tight_layout()
plt.savefig('planteamiento_trama.png', dpi=100)
plt.close()
```
