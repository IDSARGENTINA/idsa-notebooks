# Relación datos-tinta y claridad

El caso conductor de la unidad es una encuesta de satisfacción de clientes de una cadena de supermercados, con datos de producto, trimestre, satisfacción, ventas y número de clientes. A partir de esta tabla se plantean preguntas analíticas que guían la elección del gráfico y la aplicación de los principios de percepción, claridad y honestidad visual.


## Proporción tinta-datos

La proporción tinta-datos es un concepto que distingue entre la tinta dedicada a los datos y la que no lo está, y que se define como la proporción de la tinta de un gráfico dedicada a la exhibición no redundante de información de datos. La recomendación de maximizar esta proporción debe interpretarse con moderación: si se entiende como eliminar el desorden y buscar diseños limpios y elegantes, es un consejo razonable; pero si se lleva al extremo de eliminar toda la tinta que no son datos, se producen decisiones de diseño pobres. Tanto el exceso de minimalismo como el exceso de elementos recargados conducen a figuras poco atractivas. Entre los extremos existe una amplia gama de diseños aceptables que pueden ser apropiados en diferentes contextos.

**Ejemplo:** En la encuesta de satisfacción de clientes de la cadena de supermercados, al construir un gráfico de barras con matplotlib para comparar la satisfacción promedio por trimestre, se puede aplicar la proporción tinta-datos eliminando líneas de cuadrícula innecesarias y colores de fondo, pero conservando las etiquetas de los ejes y los valores, para lograr un diseño limpio sin caer en un minimalismo que dificulte la lectura.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos de la encuesta de satisfacción por trimestre
datos = pd.DataFrame({
    'trimestre': ['T1', 'T2', 'T3', 'T4'],
    'satisfaccion': [7.2, 7.8, 8.1, 8.5]
})

# Crear la figura y el eje
fig, ax = plt.subplots(figsize=(8, 5))

# Gráfico de barras con un color limpio y sin bordes innecesarios
barras = ax.bar(datos['trimestre'], datos['satisfaccion'], color='#4C72B0', edgecolor='none')

# Eliminar las líneas de cuadrícula y los bordes del gráfico para reducir tinta no dedicada a los datos
ax.grid(False)
for spine in ['top', 'right', 'left']:
    ax.spines[spine].set_visible(False)

# Conservar las etiquetas de los ejes y los valores sobre las barras para mantener la legibilidad
ax.set_xlabel('Trimestre')
ax.set_ylabel('Satisfacción promedio')
ax.set_title('Satisfacción promedio por trimestre')
ax.set_ylim(0, 10)

# Añadir los valores numéricos encima de cada barra
for barra in barras:
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width() / 2, altura + 0.1, f'{altura:.1f}',
            ha='center', va='bottom', fontsize=10)

# Mostrar el gráfico
plt.show()
```

**Errores frecuentes:**
- Interpretar la maximización de la proporción tinta-datos como eliminar toda la tinta no dedicada a los datos, lo que lleva a diseños pobres.
- Olvidar la moderación y caer en extremos de minimalismo o recargo que producen figuras poco atractivas.

## Elementos no esenciales del gráfico

En cualquier visualización, los elementos gráficos se dividen en dos grupos: los que representan datos y los que no. Los no esenciales incluyen ejes, marcas y etiquetas de los ejes, títulos de ejes, leyendas y anotaciones. Estos elementos proporcionan contexto y estructura visual, pero no son los datos en sí. Al diseñar un gráfico conviene considerar la cantidad de tinta empleada en datos y en contexto. Una recomendación común es reducir la tinta no asociada a los datos para lograr visualizaciones menos recargadas y más elegantes. Sin embargo, el contexto y la estructura visual son importantes: minimizar en exceso estos elementos puede producir figuras difíciles de leer, confusas o poco atractivas.

**Ejemplo:** Al construir un gráfico de barras con la satisfacción promedio por trimestre, se opta por eliminar el borde superior y derecho, y por quitar la cuadrícula de fondo, dejando solo los ejes, las etiquetas y el título. Así se reduce la tinta no asociada a los datos sin perder la referencia necesaria para interpretar los valores.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos: satisfacción promedio por trimestre
datos = pd.DataFrame({
    'trimestre': ['T1', 'T2', 'T3', 'T4'],
    'satisfaccion': [7.2, 7.8, 8.1, 8.5]
})

# Creamos la figura y el eje
fig, ax = plt.subplots(figsize=(7, 5))

# Gráfico de barras con la satisfacción promedio por trimestre
ax.bar(datos['trimestre'], datos['satisfaccion'], color='#4C72B0')

# Eliminamos el borde superior y el derecho para reducir tinta no asociada a los datos
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Quitamos la cuadrícula de fondo
ax.grid(False)

# Etiquetas de los ejes y título
ax.set_xlabel('Trimestre')
ax.set_ylabel('Satisfacción promedio')
ax.set_title('Satisfacción promedio por trimestre')

# Mostramos el gráfico
plt.show()
```

**Errores frecuentes:**
- Reducir en exceso los elementos que aportan contexto y estructura visual, lo que puede generar figuras difíciles de leer o confusas.
- No considerar la cantidad de tinta dedicada a datos y a contexto al diseñar el gráfico.

## Simplificación de ejes y escalas

La simplificación de ejes y escalas consiste en ajustar la escala del eje vertical al rango real de los datos, manteniendo el máximo igual o apenas por encima del valor más alto del conjunto, y asegurando que en los gráficos de barras verticales el eje comience en cero. Además, se deben usar incrementos naturales en las escalas para facilitar la lectura. Esta práctica se complementa con la eliminación de elementos no esenciales, como cuadrículas (o su sustitución por una cuadrícula gris clara) y sombras, con el fin de enfatizar los datos y reducir los componentes gráficos que no aportan información. El objetivo es lograr una representación clara y honesta, evitando distorsiones y perspectivas falsas como el 3D, y manteniendo un factor de mentira igual a 1, de modo que el tamaño del efecto mostrado en el gráfico coincida con el tamaño del efecto en los datos.

**Ejemplo:** En la encuesta de satisfacción de clientes de la cadena de supermercados, al construir un gráfico de barras verticales para comparar la satisfacción promedio por producto, se ajusta el eje vertical para que su valor máximo sea ligeramente superior al promedio más alto registrado y se fija el cero como punto de partida. Asimismo, se eliminan las líneas de cuadrícula y las sombras, dejando solo las barras y las etiquetas necesarias, con lo cual se destacan los datos de satisfacción y se reduce el ruido visual.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Datos ilustrativos: satisfacción promedio por producto (escala de 0 a 10)
datos = pd.DataFrame({
    'producto': ['Lácteos', 'Panadería', 'Bebidas', 'Carnes', 'Frutas y Verduras'],
    'satisfaccion': [8.4, 7.9, 8.8, 7.5, 9.1]
})

# Ajustamos el eje vertical: comienza en cero y el máximo queda apenas por encima del valor más alto
valor_maximo = datos['satisfaccion'].max()
limite_superior = np.ceil(valor_maximo) + 0.5  # máximo ligeramente superior al dato más alto

fig, ax = plt.subplots(figsize=(8, 5))

# Barras verticales sin sombras ni bordes que agreguen ruido visual
ax.bar(datos['producto'], datos['satisfaccion'], color='#4C72B0', edgecolor='none')

# Eje vertical desde cero hasta el límite ajustado, con incrementos naturales de 1
ax.set_ylim(0, limite_superior)
ax.set_yticks(np.arange(0, limite_superior + 0.1, 1))

# Eliminamos la cuadrícula y los marcos no esenciales
ax.grid(False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Etiquetas y título necesarios para interpretar los datos
ax.set_xlabel('Producto')
ax.set_ylabel('Satisfacción promedio (0 a 10)')
ax.set_title('Satisfacción promedio por producto')

# Mostramos el valor sobre cada barra para enfatizar los datos
for i, valor in enumerate(datos['satisfaccion']):
    ax.text(i, valor + 0.1, f'{valor:.1f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

**Errores frecuentes:**
- No ajustar el máximo del eje y al valor más alto de los datos, dejando espacio vacío innecesario.
- No asegurar que el eje y de los gráficos de barras verticales comience en cero.
- Usar incrementos no naturales en las escalas.
- Incluir cuadrículas oscuras o elementos no esenciales que dificulten la lectura de los datos.
- Agregar sombras o perspectivas falsas como el 3D que distorsionan la percepción.

## Codificación visual eficiente

Una codificación visual eficiente busca que la figura se interprete sin necesidad de una leyenda. Aunque la legibilidad de una leyenda puede mejorarse codificando los datos de forma redundante en varias estéticas, la leyenda siempre impone una carga mental adicional: el lector debe tomar la información de una parte de la visualización y transferirla a otra. Por eso, conviene eliminar la leyenda, pero no basta con omitirla y escribir en el pie de figura frases como "los puntos amarillos representan tal categoría". Eliminar la leyenda significa diseñar la figura de modo que sea inmediatamente obvio qué representa cada elemento gráfico, incluso sin leyenda explícita. La estrategia general para lograrlo es el etiquetado directo, que consiste en colocar etiquetas de texto u otros elementos visuales que sirvan de guía para el resto de la figura.

**Ejemplo:** Para representar la satisfacción promedio por producto y trimestre, en lugar de agregar una leyenda con los nombres de los productos, se escribe el nombre de cada producto al final de su línea o junto a su punto, de modo que cada elemento quede identificado sin leyenda.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos: satisfacción promedio por producto y trimestre
datos = {
    'Trimestre': ['T1', 'T2', 'T3', 'T4'],
    'Producto A': [7.2, 7.8, 8.1, 8.5],
    'Producto B': [6.5, 6.9, 7.4, 7.8],
    'Producto C': [8.0, 7.6, 7.9, 8.2]
}
df = pd.DataFrame(datos)

# Crear la figura
plt.figure(figsize=(8, 5))

# Dibujar una línea por producto
plt.plot(df['Trimestre'], df['Producto A'], marker='o', color='tab:blue')
plt.plot(df['Trimestre'], df['Producto B'], marker='o', color='tab:orange')
plt.plot(df['Trimestre'], df['Producto C'], marker='o', color='tab:green')

# Etiquetado directo: escribir el nombre de cada producto al final de su línea
plt.text(df['Trimestre'].iloc[-1], df['Producto A'].iloc[-1], '  Producto A',
         va='center', ha='left', color='tab:blue', fontweight='bold')
plt.text(df['Trimestre'].iloc[-1], df['Producto B'].iloc[-1], '  Producto B',
         va='center', ha='left', color='tab:orange', fontweight='bold')
plt.text(df['Trimestre'].iloc[-1], df['Producto C'].iloc[-1], '  Producto C',
         va='center', ha='left', color='tab:green', fontweight='bold')

# Títulos y etiquetas de ejes
plt.title('Satisfacción promedio por producto y trimestre')
plt.xlabel('Trimestre')
plt.ylabel('Satisfacción promedio')

# Ajustar el margen derecho para que las etiquetas se vean completas
plt.subplots_adjust(right=0.8)

# Mostrar el gráfico sin leyenda
plt.show()
```

**Errores frecuentes:**
- Omitir la leyenda y reemplazarla por una frase en el pie de figura que explique qué representa cada elemento.
- Mantener la leyenda aunque se codifiquen los datos de forma redundante en varias estéticas.

## Contraste y jerarquía visual

El contraste es una señal que orienta la atención del público hacia lo que se considera relevante. Cuando se introducen muchas diferencias visuales a la vez, cada una compite con las demás y ninguna logra destacar; la falta de un contraste claro produce caos visual. La analogía del halcón entre palomas ilustra que distinguir un elemento singular es fácil, pero a medida que aumenta la diversidad de aves, la identificación se vuelve difícil. Por eso, el uso estratégico del contraste implica reservar la diferencia visual más marcada para el elemento realmente importante, de modo que sea lo único que se distinga claramente del resto.

**Ejemplo:** En el tablero de la encuesta de satisfacción, si se colorean de rojo todos los trimestres con satisfacción baja y además se agregan íconos y etiquetas de colores, el espectador no sabrá dónde mirar. En cambio, si solo se resalta con un color intenso el trimestre con menor satisfacción y se dejan los demás en gris, ese dato se convierte en el halcón y capta la atención de inmediato.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos de la encuesta de satisfacción por trimestre
datos = pd.DataFrame({
    'trimestre': ['T1', 'T2', 'T3', 'T4'],
    'satisfaccion': [72, 65, 58, 80]
})

# Identificamos el trimestre con menor satisfacción
indice_min = datos['satisfaccion'].idxmin()

# Colores: gris para todos, rojo intenso solo para el mínimo
colores = ['#b0b0b0'] * len(datos)
colores[indice_min] = '#d62728'

# Gráfico de barras con contraste jerárquico
fig, ax = plt.subplots(figsize=(8, 5))
barras = ax.bar(datos['trimestre'], datos['satisfaccion'], color=colores)

# Etiquetas y título
ax.set_title('Satisfacción por trimestre: resaltando el mínimo', fontsize=14)
ax.set_xlabel('Trimestre')
ax.set_ylabel('Satisfacción (%)')

# Anotamos el valor del trimestre resaltado
ax.text(indice_min, datos['satisfaccion'][indice_min] + 2,
        f"{datos['satisfaccion'][indice_min]}%",
        ha='center', color='#d62728', fontweight='bold')

plt.show()
```

**Errores frecuentes:**
- Introducir demasiadas diferencias visuales a la vez, de modo que ningún elemento destaque.
- No reservar el contraste más marcado para el dato realmente importante.

### Solución


```python
import matplotlib.pyplot as plt

# Datos de satisfacción promedio por trimestre
trimestres = ['T1', 'T2', 'T3', 'T4']
satisfaccion = [7.2, 6.8, 8.1, 7.5]

fig, ax = plt.subplots()

# Identificamos el trimestre con menor satisfacción
min_idx = satisfaccion.index(min(satisfaccion))
colores = ['lightgray'] * len(satisfaccion)
colores[min_idx] = 'crimson'

ax.bar(trimestres, satisfaccion, color=colores, edgecolor='none')
ax.set_title('Satisfacción promedio por trimestre')
ax.set_xlabel('Trimestre')
ax.set_ylabel('Satisfacción promedio')

# Eliminamos cuadrícula y bordes superior e derecho
ax.grid(False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Ajustamos el eje y para que comience en cero y el máximo sea apenas superior al valor más alto
ax.set_ylim(0, max(satisfaccion) + 0.5)

plt.show()
```
