# Jerarquía y composición

El caso conductor es un dashboard ejecutivo de una cadena de supermercados con datos de ventas, satisfacción, clientes y margen por producto y trimestre, para responder preguntas de negocio y diseñar un panel accesible y jerárquico.


## Peso visual y punto focal

La ausencia de un contraste claro puede generar caos visual. El contraste se presenta como una señal para orientar la atención del público hacia lo relevante. La analogía del halcón entre palomas ilustra que, cuando hay mucha diversidad de elementos, ninguno se distingue; por eso, si algo es realmente importante (el halcón), debe ser lo único muy diferente del resto. Así, el uso estratégico del contraste permite que el punto focal emerja y evita la confusión.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados, se muestran ventas, satisfacción, clientes y margen por producto y trimestre. Si todas las métricas tienen el mismo color, tamaño y saturación, el directivo no sabe dónde mirar primero. Para evitar el caos visual, se podría resaltar solo la métrica de margen con un color intenso y un tamaño mayor, mientras las demás quedan en tonos neutros. Así, el margen se convierte en el halcón y guía la atención hacia la pregunta de negocio prioritaria. (Datos ilustrativos).

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard ejecutivo de una cadena de supermercados
datos = {
    'Métrica': ['Ventas', 'Satisfacción', 'Clientes', 'Margen'],
    'Valor': [85, 72, 90, 45]
}
df = pd.DataFrame(datos)

# Creamos dos paneles: uno con caos visual (todo igual) y otro con punto focal
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# --- Panel 1: Caos visual, todas las métricas con el mismo color y tamaño ---
colores_caos = ['#7f7f7f'] * len(df)
ax1.bar(df['Métrica'], df['Valor'], color=colores_caos)
ax1.set_title('Sin contraste: caos visual')
ax1.set_ylabel('Valor')
ax1.set_ylim(0, 100)

# --- Panel 2: Punto focal, solo el Margen resaltado ---
# El margen es el 'halcón': color intenso y mayor tamaño
colores_foco = ['#cccccc', '#cccccc', '#cccccc', '#d62728']
tamanos = [1, 1, 1, 1.6]  # el margen se dibuja más ancho
barras = ax2.bar(df['Métrica'], df['Valor'], color=colores_foco)
# Ajustamos el ancho de la barra del margen para que sea más grande
barras[3].set_width(1.6)
ax2.set_title('Con contraste: el margen es el punto focal')
ax2.set_ylabel('Valor')
ax2.set_ylim(0, 100)

# Etiquetamos el valor del margen para reforzar la atención
ax2.annotate('Punto focal', xy=(3, 45), xytext=(2.2, 70),
             arrowprops=dict(arrowstyle='->', color='#d62728'),
             color='#d62728', fontweight='bold')

plt.tight_layout()
plt.savefig('peso_visual_punto_focal.png')
```

**Errores frecuentes:**
- No usar estratégicamente el contraste, lo que impide que algún elemento destaque.
- Hacer muchas cosas diferentes sin jerarquía, reduciendo la posibilidad de que algo resalte.

## Agrupacion por proximidad y espacio

Conocer estos principios es muy útil para crear cualquier método de visualización. Se ilustra con el principio de proximidad, que se refiere al enfoque visual de agrupar formas si se ven similares entre sí. Un grupo de este tipo suele percibirse como una única unidad. A continuación, anuncia una imagen que muestra cómo se puede distinguir la proximidad.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados (datos ilustrativos), se colocan juntas las métricas de ventas y margen por producto, separadas de las de satisfacción y clientes. Así, la proximidad sugiere que las primeras forman un grupo de indicadores financieros y las segundas un grupo de indicadores de experiencia, guiando la interpretación del usuario.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard ejecutivo (cadena de supermercados)
datos = pd.DataFrame({
    'metrica': ['Ventas', 'Margen', 'Satisfaccion', 'Clientes'],
    'valor': [120, 35, 88, 450],
    'grupo': ['Financiero', 'Financiero', 'Experiencia', 'Experiencia']
})

# Colores por grupo: mismo color para metricas cercanas (proximidad)
colores = {'Financiero': '#1f77b4', 'Experiencia': '#ff7f0e'}

fig, ax = plt.subplots(figsize=(9, 5))

# Posiciones: Ventas y Margen juntas (izquierda), Satisfaccion y Clientes juntas (derecha)
# El espacio grande entre ambos pares refuerza la agrupacion por proximidad
posiciones = [0, 1, 3.5, 4.5]
barras = ax.bar(posiciones, datos['valor'],
                color=[colores[g] for g in datos['grupo']], width=0.8)

# Etiquetas de cada metrica
ax.set_xticks(posiciones)
ax.set_xticklabels(datos['metrica'])

# Titulo y etiquetas
ax.set_title('Dashboard ejecutivo: agrupacion por proximidad', fontsize=13)
ax.set_ylabel('Valor')

# Leyenda que explica los dos grupos formados por cercania
from matplotlib.patches import Patch
handles = [Patch(color=colores['Financiero'], label='Indicadores financieros'),
           Patch(color=colores['Experiencia'], label='Indicadores de experiencia')]
ax.legend(handles=handles, loc='upper right')

# Anotaciones que marcan visualmente los dos grupos
ax.annotate('Grupo financiero', xy=(0.5, 130), ha='center', fontsize=10, color='#1f77b4')
ax.annotate('Grupo experiencia', xy=(4.0, 470), ha='center', fontsize=10, color='#ff7f0e')

plt.tight_layout()
plt.savefig('proximidad_dashboard.png', dpi=100)
print('Figura guardada: proximidad_dashboard.png')
```

## Retícula y estructura de composicion

La retícula y estructura de composición son fundamentales en el diseño de dashboards profesionales, ya que los sistemas de cuadrícula proporcionan una base estructural para organizar múltiples gráficos y mantener el equilibrio visual y la consistencia. Según, estos sistemas aseguran coherencia visual entre diferentes tipos y tamaños de gráficos, y facilitan la comparación entre métricas relacionadas. Añade que los layouts más efectivos siguen patrones establecidos que optimizan la presentación de información para distintos contextos analíticos, como la pirámide invertida, el cuarteto de dashboard o el resumen ejecutivo.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados (datos ilustrativos), se aplica una cuadrícula de 2x2 para comparar ventas, satisfacción, clientes y margen por producto y trimestre. Esta estructura permite una comparación rápida y equilibrada entre los cuatro indicadores clave, facilitando la identificación de tendencias y la toma de decisiones.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard ejecutivo de una cadena de supermercados
datos = {
    'Trimestre': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Ventas': [120, 135, 150, 160],
    'Satisfaccion': [4.2, 4.3, 4.5, 4.6],
    'Clientes': [1000, 1100, 1200, 1300],
    'Margen': [0.25, 0.26, 0.27, 0.28]
}
df = pd.DataFrame(datos)

# Crear una figura con una cuadrícula de 2x2 (retícula de composición)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
fig.suptitle('Dashboard Ejecutivo - Cadena de Supermercados', fontsize=16)

# Gráfico 1: Ventas por trimestre (barras)
axes[0, 0].bar(df['Trimestre'], df['Ventas'], color='skyblue')
axes[0, 0].set_title('Ventas por Trimestre')
axes[0, 0].set_ylabel('Ventas (miles)')

# Gráfico 2: Satisfacción por trimestre (línea)
axes[0, 1].plot(df['Trimestre'], df['Satisfaccion'], marker='o', color='green')
axes[0, 1].set_title('Satisfacción por Trimestre')
axes[0, 1].set_ylabel('Satisfacción (1-5)')

# Gráfico 3: Clientes por trimestre (barras)
axes[1, 0].bar(df['Trimestre'], df['Clientes'], color='orange')
axes[1, 0].set_title('Clientes por Trimestre')
axes[1, 0].set_ylabel('Número de clientes')

# Gráfico 4: Margen por trimestre (línea)
axes[1, 1].plot(df['Trimestre'], df['Margen'], marker='s', color='red')
axes[1, 1].set_title('Margen por Trimestre')
axes[1, 1].set_ylabel('Margen (%)')

# Ajustar el espaciado para mantener una composición equilibrada
plt.tight_layout()

# Guardar la figura (no mostrar en pantalla)
plt.savefig('dashboard_reticula_2x2.png')
```

## Orden de lectura y flujo visual

El concepto de orden de lectura y flujo visual se refiere a la importancia de alinear la presentación de la información con la forma natural en que el público la procesa. Literal, 'He aquí un ejemplo de lo que supone pedir al público que trabaje a contracorriente: en cierta ocasión,', introduce un caso en el que un diagrama de flujo comenzaba en la parte inferior derecha y debía leerse hacia arriba y hacia la izquierda, lo que resultó incómodo para el autor. Enfatiza que se debe evitar que la audiencia sienta incomodidad al interpretar una visualización. En contraste, el orden natural de lectura suele ser de arriba a la izquierda hacia abajo a la derecha, por lo que diseñar en contra de esta expectativa obliga al público a un esfuerzo adicional. Otro ejemplo mencionado es una escala de valores en la que los positivos se colocan a la izquierda y los negativos a la derecha, lo que contradice asociaciones naturales y dificulta la comprensión. Por lo tanto, el orden de lectura y el flujo visual deben respetar las convenciones para facilitar el procesamiento de la información.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados, se podría organizar el flujo visual de manera que los indicadores clave de ventas y satisfacción se ubiquen en la esquina superior izquierda y se avance hacia la derecha y hacia abajo, siguiendo el orden natural de lectura. Si en cambio se colocaran los datos más relevantes en la parte inferior derecha y se pidiera al usuario que los buscara allí primero, se estaría yendo a contracorriente, como en el ejemplo del diagrama de flujo mencionado. Datos ilustrativos: ventas trimestrales, satisfacción del cliente, número de clientes y margen por producto.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard ejecutivo
datos = {
    'Indicador': ['Ventas trimestrales', 'Satisfacción del cliente', 'Número de clientes', 'Margen por producto'],
    'Valor': [1200000, 85, 45000, 32]
}
df = pd.DataFrame(datos)

# Creamos una figura con dos subplots: uno con flujo natural y otro a contracorriente
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# --- Subplot izquierdo: flujo natural (arriba-izquierda a abajo-derecha) ---
ax = axes[0]
ax.set_title('Flujo natural: de arriba-izquierda a abajo-derecha', fontsize=12)
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')  # Ocultamos ejes para simular un dashboard

# Posiciones de los indicadores en orden de lectura natural
posiciones_natural = [
    (1, 8, 'Ventas trimestrales\n$1,200,000'),
    (5, 8, 'Satisfacción del cliente\n85%'),
    (1, 4, 'Número de clientes\n45,000'),
    (5, 4, 'Margen por producto\n32%')
]
for x, y, texto in posiciones_natural:
    ax.text(x, y, texto, ha='center', va='center', fontsize=10,
            bbox=dict(boxstyle='round,pad=0.5', facecolor='lightblue', edgecolor='blue'))

# Flechas que indican el flujo de lectura
ax.annotate('', xy=(4.5, 8), xytext=(2.5, 8), arrowprops=dict(arrowstyle='->', color='gray'))
ax.annotate('', xy=(1, 5.5), xytext=(1, 7), arrowprops=dict(arrowstyle='->', color='gray'))
ax.annotate('', xy=(4.5, 4), xytext=(2.5, 4), arrowprops=dict(arrowstyle='->', color='gray'))

# --- Subplot derecho: flujo a contracorriente (abajo-derecha hacia arriba-izquierda) ---
ax = axes[1]
ax.set_title('Flujo a contracorriente: de abajo-derecha a arriba-izquierda', fontsize=12)
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Posiciones de los indicadores en orden inverso (contracorriente)
posiciones_contra = [
    (8, 2, 'Ventas trimestrales\n$1,200,000'),
    (4, 2, 'Satisfacción del cliente\n85%'),
    (8, 6, 'Número de clientes\n45,000'),
    (4, 6, 'Margen por producto\n32%')
]
for x, y, texto in posiciones_contra:
    ax.text(x, y, texto, ha='center', va='center', fontsize=10,
            bbox=dict(boxstyle='round,pad=0.5', facecolor='lightcoral', edgecolor='red'))

# Flechas que indican el flujo de lectura inverso
ax.annotate('', xy=(5.5, 2), xytext=(7.5, 2), arrowprops=dict(arrowstyle='->', color='gray'))
ax.annotate('', xy=(8, 4.5), xytext=(8, 3), arrowprops=dict(arrowstyle='->', color='gray'))
ax.annotate('', xy=(5.5, 6), xytext=(7.5, 6), arrowprops=dict(arrowstyle='->', color='gray'))

plt.tight_layout()
plt.savefig('flujo_visual.png')
plt.close()
```

**Errores frecuentes:**
- Comenzar un diagrama de flujo en la parte inferior derecha y hacer que se lea hacia arriba y hacia la izquierda, lo que resulta incómodo para el público.
- Representar una escala de valores negativos a positivos con los valores positivos a la izquierda y los negativos a la derecha, contradiciendo asociaciones naturales y dificultando la interpretación.

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos ilustrativos: ventas, satisfacción, clientes y margen por trimestre
trimestres = ['Q1', 'Q2', 'Q3', 'Q4']
ventas = [120, 135, 150, 160]
satisfaccion = [80, 82, 85, 88]
clientes = [200, 210, 220, 230]
margen = [15, 18, 20, 22]

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Grupo financiero: ventas y margen juntos (arriba-izquierda y arriba-derecha)
# Margen resaltado como punto focal: color intenso y mayor tamaño
axs[0, 0].plot(trimestres, ventas, marker='o', color='gray', linewidth=1)
axs[0, 0].set_title('Ventas')
axs[0, 1].plot(trimestres, margen, marker='o', color='red', linewidth=3, markersize=10)
axs[0, 1].set_title('Margen', fontsize=14, fontweight='bold')

# Grupo de experiencia: satisfacción y clientes (abajo-izquierda y abajo-derecha)
axs[1, 0].plot(trimestres, satisfaccion, marker='o', color='gray', linewidth=1)
axs[1, 0].set_title('Satisfacción')
axs[1, 1].plot(trimestres, clientes, marker='o', color='gray', linewidth=1)
axs[1, 1].set_title('Clientes')

# Ajustar espacio para agrupar visualmente por proximidad
plt.subplots_adjust(wspace=0.4, hspace=0.4)
plt.savefig('practica.png')
```
