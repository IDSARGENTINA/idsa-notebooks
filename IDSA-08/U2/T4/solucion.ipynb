## Contraste de color suficiente

El contraste de color es un aspecto crítico para que las visualizaciones de datos sean accesibles a usuarios con baja visión. Según, una visualización de datos debe tener suficiente contraste entre el texto y el fondo. El contraste se mide por la diferencia de luminancia entre dos colores, y que el estándar WCAG 2.1 AA establece relaciones de contraste mínimas: al menos 3:1 para texto grande o en negrita e iconos, y 4.5:1 para texto de lectura general. Además, se mencionan herramientas como el Analizador de Contraste de Color de TPGi y la configuración de modo de alto contraste del sistema operativo para verificar y ajustar el contraste.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados (datos ilustrativos), se muestra un gráfico de barras con etiquetas de texto gris claro sobre fondo blanco. Al medir la relación de contraste con una herramienta, se obtiene 2.5:1, lo cual está por debajo del mínimo de 4.5:1 para texto de lectura general según WCAG 2.1 AA. Para corregirlo, se cambia el color del texto a un gris más oscuro que alcance al menos 4.5:1, garantizando la legibilidad para usuarios con baja visión.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Datos ilustrativos del dashboard ejecutivo de una cadena de supermercados
datos = pd.DataFrame({
    'Sucursal': ['Norte', 'Sur', 'Centro', 'Este', 'Oeste'],
    'Ventas': [120, 95, 140, 80, 110]
})

# Función para calcular la luminancia relativa de un color (WCAG)
def luminancia(hex_color):
    hex_color = hex_color.lstrip('#')
    r, g, b = [int(hex_color[i:i+2], 16) / 255 for i in (0, 2, 4)]
    def canal(c):
        return c / 12.92 if c <= 0.03928 else ((c + 0.055) / 1.055) ** 2.4
    r, g, b = canal(r), canal(g), canal(b)
    return 0.2126 * r + 0.7152 * g + 0.0722 * b

# Función para calcular la relación de contraste entre dos colores
def relacion_contraste(color1, color2):
    l1, l2 = luminancia(color1), luminancia(color2)
    claro, oscuro = max(l1, l2), min(l1, l2)
    return (claro + 0.05) / (oscuro + 0.05)

# Colores de texto: gris claro (mal contraste) y gris oscuro (buen contraste)
gris_claro = '#B0B0B0'
gris_oscuro = '#595959'
fondo = '#FFFFFF'

contraste_claro = relacion_contraste(gris_claro, fondo)
contraste_oscuro = relacion_contraste(gris_oscuro, fondo)
print(f'Contraste gris claro sobre blanco: {contraste_claro:.2f}:1')
print(f'Contraste gris oscuro sobre blanco: {contraste_oscuro:.2f}:1')

# Crear figura con dos subgráficos: antes y después de la corrección
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# --- Gráfico ANTES: etiquetas gris claro (contraste insuficiente) ---
ax = axes[0]
ax.bar(datos['Sucursal'], datos['Ventas'], color='#4C72B0')
ax.set_title('ANTES: texto gris claro (contraste insuficiente)', fontsize=12)
ax.set_ylabel('Ventas (miles)', color=gris_claro)
ax.tick_params(axis='x', colors=gris_claro)
ax.tick_params(axis='y', colors=gris_claro)
for spine in ax.spines.values():
    spine.set_color(gris_claro)
# Etiquetas de valor sobre las barras en gris claro
for i, v in enumerate(datos['Ventas']):
    ax.text(i, v + 2, str(v), ha='center', color=gris_claro, fontsize=10)
ax.text(0.5, -0.18, f'Relación de contraste: {contraste_claro:.2f}:1 (mínimo 4.5:1)',
        transform=ax.transAxes, ha='center', color='red', fontsize=10)

# --- Gráfico DESPUÉS: etiquetas gris oscuro (contraste suficiente) ---
ax = axes[1]
ax.bar(datos['Sucursal'], datos['Ventas'], color='#4C72B0')
ax.set_title('DESPUÉS: texto gris oscuro (contraste suficiente)', fontsize=12)
ax.set_ylabel('Ventas (miles)', color=gris_oscuro)
ax.tick_params(axis='x', colors=gris_oscuro)
ax.tick_params(axis='y', colors=gris_oscuro)
for spine in ax.spines.values():
    spine.set_color(gris_oscuro)
for i, v in enumerate(datos['Ventas']):
    ax.text(i, v + 2, str(v), ha='center', color=gris_oscuro, fontsize=10)
ax.text(0.5, -0.18, f'Relación de contraste: {contraste_oscuro:.2f}:1 (cumple 4.5:1)',
        transform=ax.transAxes, ha='center', color='green', fontsize=10)

plt.tight_layout()
plt.savefig('contraste_color.png', dpi=100)
print('Figura guardada como contraste_color.png')
```

## No depender solo del color

Es muy difícil encontrar un conjunto de colores donde todos tengan una relación de contraste de 3:1, incluso en gráficos con solo dos categorías o series. Esto implica que no se debe depender únicamente del color para distinguir categorías, ya que el contraste entre colores puede ser insuficiente. Recomienda elegir un conjunto monocromático con el mayor contraste posible, que podría ser tan bajo como 1.3:1, y sugiere usar espaciado, posición, etiquetas o estilos interactivos y tooltips como otras formas de indicar a qué categoría pertenece cada dato. Además, aunque los patrones se recomiendan ampliamente para proporcionar contraste visual, deben usarse con cuidado porque pueden hacer que el gráfico sea menos accesible al ser visualmente confusos para algunos lectores. Por lo tanto, la idea central es que el color por sí solo no es suficiente para garantizar la accesibilidad; se deben emplear múltiples canales de codificación.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados (datos ilustrativos), se muestran las ventas por trimestre para dos categorías: 'Alimentos' y 'Bebidas'. Si se usan dos tonos de azul con una relación de contraste de 1.5:1, los usuarios con baja visión podrían no distinguirlos. Para evitar depender solo del color, se añaden etiquetas de texto directamente sobre las barras y se separan las barras con espaciado. Así, incluso si los colores no son distinguibles, la información sigue siendo accesible.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos: ventas por trimestre para dos categorías
datos = {
    'Trimestre': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Alimentos': [120, 150, 130, 170],
    'Bebidas': [80, 95, 110, 105]
}
df = pd.DataFrame(datos)

# Crear figura y ejes
fig, ax = plt.subplots(figsize=(8, 5))

# Posiciones de las barras: separar las barras de cada categoría con espaciado
x = range(len(df['Trimestre']))
ancho = 0.35
pos_alimentos = [i - ancho/2 for i in x]
pos_bebidas = [i + ancho/2 for i in x]

# Colores: dos tonos de azul con bajo contraste (simulando 1.5:1)
color_alimentos = '#6baed6'  # azul claro
color_bebidas = '#4292c6'    # azul medio (contraste bajo con el anterior)

# Dibujar barras
bars1 = ax.bar(pos_alimentos, df['Alimentos'], width=ancho, label='Alimentos', color=color_alimentos)
bars2 = ax.bar(pos_bebidas, df['Bebidas'], width=ancho, label='Bebidas', color=color_bebidas)

# Añadir etiquetas de texto directamente sobre las barras para no depender solo del color
for bar in bars1:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 2, f'{height}', ha='center', va='bottom', fontsize=9)
for bar in bars2:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 2, f'{height}', ha='center', va='bottom', fontsize=9)

# Configurar ejes y título
ax.set_xticks(x)
ax.set_xticklabels(df['Trimestre'])
ax.set_ylabel('Ventas (miles de unidades)')
ax.set_title('Ventas por trimestre - Alimentos vs Bebidas')
ax.legend()

# Ajustar límites para dar espacio a las etiquetas
ax.set_ylim(0, max(df['Alimentos'].max(), df['Bebidas'].max()) * 1.15)

# Guardar la figura (sin mostrar)
plt.tight_layout()
plt.savefig('ventas_trimestre_accesible.png')
plt.close()
```

**Errores frecuentes:**
- Depender únicamente del color para distinguir categorías cuando el contraste entre colores es bajo.
- Usar patrones sin cuidado, ya que pueden hacer que el gráfico sea menos accesible al ser visualmente confusos para algunos lectores.

## Etiquetas y texto alternativo en graficos

Los metodos innovadores de visualizacion de datos aparecen cada vez mas en diversos medios, como sitios web de noticias, portales de informacion gubernamental, estados financieros y comunicacion cientifica, frecuentemente sin descripcion para usuarios de tecnologia de lectura de pantalla. Esto implica que muchas visualizaciones carecen de etiquetas o texto alternativo que las hagan accesibles. Anade que, incluso cuando se proporciona texto alternativo, a menudo no logra transmitir las conclusiones principales de los datos. Ademas, para visualizaciones producidas como archivos graficos (JPG, PDF, PNG), es necesario agregar texto alternativo, ya que estos archivos no pueden navegarse con software de lectura de pantalla. La mayoria del software estadistico no genera graficos accesibles, por lo que la responsabilidad recae en el autor. En sintesis, las etiquetas y el texto alternativo son esenciales para la accesibilidad, pero su implementacion es deficiente y a menudo postergada.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados (datos ilustrativos), se muestra un grafico de barras con las ventas trimestrales por producto. Sin texto alternativo, un usuario de lector de pantalla no puede acceder a la informacion. Al agregar un texto alternativo como 'Ventas del trimestre 1: producto A $120,000, producto B $95,000; el producto A lidera con un 26% mas que el B', se comunican las conclusiones principales. Sin embargo, si el texto alternativo solo dijera 'Grafico de ventas', no cumpliria su funcion. Este ejemplo ilustra la necesidad de etiquetas descriptivas y significativas.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard ejecutivo de una cadena de supermercados
datos = pd.DataFrame({
    'Producto': ['Producto A', 'Producto B'],
    'Ventas_Trimestre1': [120000, 95000]
})

# Crear figura y eje 2D
fig, ax = plt.subplots(figsize=(8, 5))

# Grafico de barras con las ventas trimestrales por producto
barras = ax.bar(datos['Producto'], datos['Ventas_Trimestre1'],
                color=['#1f77b4', '#ff7f0e'])

# Etiquetas de ejes y titulo descriptivo
ax.set_xlabel('Producto')
ax.set_ylabel('Ventas (USD)')
ax.set_title('Ventas del Trimestre 1 por Producto')

# Etiquetas de valor sobre cada barra para que la informacion sea legible
for barra in barras:
    altura = barra.get_height()
    ax.annotate(f'${altura:,.0f}',
                xy=(barra.get_x() + barra.get_width() / 2, altura),
                xytext=(0, 3), textcoords='offset points',
                ha='center', va='bottom')

# Texto alternativo significativo: comunica las conclusiones principales
# (no solo 'Grafico de ventas', sino los valores y la comparacion)
texto_alternativo = (
    'Ventas del trimestre 1: producto A $120,000, producto B $95,000; '
    'el producto A lidera con un 26% mas que el B'
)
fig.text(0.5, 0.01, 'Texto alternativo: ' + texto_alternativo,
         ha='center', va='bottom', fontsize=8, wrap=True)

# Ajustar layout y guardar (sin plt.show por falta de display)
plt.tight_layout(rect=[0, 0.08, 1, 1])
plt.savefig('ventas_trimestre1_accesible.png', dpi=100)
plt.close(fig)
```

**Errores frecuentes:**
- Proporcionar texto alternativo que no logra transmitir las conclusiones principales de los datos.
- No agregar texto alternativo a visualizaciones en archivos graficos como JPG, PDF o PNG.
- Asumir que el software estadistico genera graficos accesibles automaticamente, cuando la mayoria no lo hace.

## Jerarquia visual accesible

La jerarquía visual accesible implica que el diseño de un gráfico debe organizar los elementos de manera que el público pueda procesar la información sin que parezca complicado. Esta jerarquía se logra destacando lo importante, eliminando distracciones y estableciendo un orden visual meditado. Así, el público utiliza esa jerarquía para recorrer y comprender el gráfico de forma intuitiva. Además, el concepto de accesibilidad amplía esta idea: los diseños deben poder ser usados por personas con diferentes capacidades, no solo aquellas con discapacidades, sino en un sentido más general. Por lo tanto, una jerarquía visual accesible combina la organización clara de la información con la usabilidad para diversos usuarios, facilitando la comprensión de los elementos visuales.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados (datos ilustrativos), se podría aplicar una jerarquía visual accesible colocando el indicador de ventas totales en la parte superior con mayor tamaño y color contrastante, mientras que los detalles por producto y trimestre se ubican en segundo plano con menor énfasis. Se eliminarían elementos decorativos innecesarios y se usarían etiquetas claras y tamaños de fuente legibles para personas con baja visión. Así, el usuario puede identificar rápidamente lo más relevante y profundizar según necesite, sin sentirse abrumado.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard ejecutivo de una cadena de supermercados
ventas_totales = 1250000  # indicador principal (ventas totales)
detalle = pd.DataFrame({
    'Producto': ['Lácteos', 'Carnes', 'Frutas', 'Bebidas'],
    'Ventas': [320000, 410000, 250000, 270000]
})
trimestres = pd.DataFrame({
    'Trimestre': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Ventas': [280000, 310000, 330000, 330000]
})

# Creamos una figura con jerarquía visual: el indicador principal arriba y grande,
# los detalles en segundo plano con menor énfasis
fig = plt.figure(figsize=(10, 8))

# --- Indicador principal (ventas totales) en la parte superior, mayor tamaño y color contrastante ---
ax_top = fig.add_axes([0.1, 0.72, 0.8, 0.22])
ax_top.axis('off')  # sin ejes para destacar solo el número
ax_top.text(0.5, 0.6, f'${ventas_totales:,.0f}',
            ha='center', va='center', fontsize=48, fontweight='bold', color='#1f4e79')
ax_top.text(0.5, 0.15, 'Ventas Totales',
            ha='center', va='center', fontsize=18, color='#1f4e79')

# --- Detalle por producto (segundo plano, menor énfasis) ---
ax_prod = fig.add_axes([0.08, 0.08, 0.38, 0.45])
ax_prod.bar(detalle['Producto'], detalle['Ventas'], color='#a6bddb')
ax_prod.set_title('Ventas por Producto', fontsize=12, color='#555555')
ax_prod.tick_params(labelsize=10)
ax_prod.set_ylabel('Ventas', fontsize=10)
# Eliminamos elementos decorativos innecesarios (bordes superior y derecho)
ax_prod.spines['top'].set_visible(False)
ax_prod.spines['right'].set_visible(False)

# --- Detalle por trimestre (segundo plano, menor énfasis) ---
ax_trim = fig.add_axes([0.55, 0.08, 0.38, 0.45])
ax_trim.plot(trimestres['Trimestre'], trimestres['Ventas'], marker='o', color='#a6bddb')
ax_trim.set_title('Ventas por Trimestre', fontsize=12, color='#555555')
ax_trim.tick_params(labelsize=10)
ax_trim.set_ylabel('Ventas', fontsize=10)
ax_trim.spines['top'].set_visible(False)
ax_trim.spines['right'].set_visible(False)

# Etiquetas claras y tamaños de fuente legibles para personas con baja visión
plt.savefig('dashboard_jerarquia_accesible.png', dpi=100)
plt.close(fig)
```

## Paletas seguras para daltonismo

Afirma que los colores son información y que se debe tener en cuenta la accesibilidad usando tonos contrastados para maximizar la legibilidad, además de emplear una paleta segura para daltonismo. Esto implica que la elección de colores no es solo estética, sino que transmite datos; por lo tanto, si no hay contraste suficiente o se usan combinaciones que los daltónicos no distinguen, la información se pierde. Literal enfatiza el contraste como medio para asegurar la legibilidad, y añade la necesidad de una paleta color-blind safe. En conjunto, se busca que cualquier usuario, independientemente de su percepción del color, pueda leer e interpretar la información codificada cromáticamente.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados (datos ilustrativos), se usa una paleta segura para daltonismo: azul y naranja con alto contraste para distinguir ventas y margen. Así, una persona con deuteranopía puede diferenciar ambas categorías sin confundirlas, manteniendo la legibilidad de los gráficos.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard ejecutivo de una cadena de supermercados
datos = pd.DataFrame({
    'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste'],
    'Ventas': [120, 95, 140, 80, 110],
    'Margen': [30, 22, 35, 18, 28]
})

# Paleta segura para daltonismo: azul y naranja con alto contraste
# (colores distinguibles para deuteranopía)
color_ventas = '#0072B2'  # azul
color_margen = '#E69F00'  # naranja

fig, ax = plt.subplots(figsize=(9, 5))

# Posiciones de las barras agrupadas
import numpy as np
x = np.arange(len(datos['Sucursal']))
ancho = 0.38

# Barras de ventas y margen con la paleta segura
ax.bar(x - ancho/2, datos['Ventas'], ancho, label='Ventas', color=color_ventas)
ax.bar(x + ancho/2, datos['Margen'], ancho, label='Margen', color=color_margen)

# Etiquetas y título
ax.set_xticks(x)
ax.set_xticklabels(datos['Sucursal'])
ax.set_ylabel('Miles de $')
ax.set_title('Ventas vs Margen por sucursal (paleta color-blind safe)')
ax.legend()

# Guardar la figura (sin plt.show en el sandbox)
plt.tight_layout()
plt.savefig('dashboard_paleta_segura.png', dpi=100)
plt.close()
```

```python
import matplotlib.pyplot as plt

# Datos ilustrativos de ventas por trimestre
trimestres = ['T1', 'T2', 'T3', 'T4']
alimentos = [120, 150, 130, 170]
bebidas = [95, 110, 105, 125]

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(trimestres, alimentos, label='Alimentos', color='#1f77b4')
ax.bar(trimestres, bebidas, label='Bebidas', color='#aec7e8')

# Etiquetas de datos con color de alto contraste (#333333 sobre blanco, ratio ~12:1)
for i, v in enumerate(alimentos):
    ax.text(i - 0.2, v + 2, str(v), color='#333333', fontsize=10, ha='center')
for i, v in enumerate(bebidas):
    ax.text(i + 0.2, v + 2, str(v), color='#333333', fontsize=10, ha='center')

ax.set_ylabel('Ventas (miles de $)')
ax.set_title('Ventas trimestrales por categoría')
ax.legend()

# Texto alternativo descriptivo que comunica la conclusión principal
alt_text = ('Gráfico de barras de ventas trimestrales. '

            'Alimentos supera a Bebidas en todos los trimestres; '

            'en T1, Alimentos vendió 120 mil $ frente a 95 mil $ de Bebidas.')
fig.text(0.5, 0.01, alt_text, ha='center', fontsize=8, color='#333333', wrap=True)

plt.savefig('ventas_accesible.png')
plt.close()
```
