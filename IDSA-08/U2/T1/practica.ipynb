# Color: paletas y daltonismo

El caso conductor es un dashboard ejecutivo de una cadena de supermercados con datos de ventas, satisfacción, clientes y margen por producto y trimestre, para responder preguntas de negocio y diseñar un panel accesible y jerárquico.


## Paletas accesibles para daltonismo

Indica que la paleta cualitativa presentada está pensada para todas las deficiencias en la visión del color. Proporciona, para cada color, su código hexadecimal, sus valores RGB (en escala 0-255 y en porcentaje) y sus valores CMYK. Aclara además que estos códigos pueden ingresarse directamente en muchas bibliotecas de gráficos y programas de manipulación de imágenes, y que, si el software no acepta hexadecimales, se pueden usar los valores RGB correspondientes. Finalmente, señala que la escala fue desarrollada por Okabe-Ito-CUD y la presenta como una escala amigable para personas daltónicas (colorblind-friendly).

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados, se asignan los colores de la paleta Okabe-Ito a cuatro categorías de productos: naranja (#E69F00) para bebidas, azul cielo (#56B4E9) para lácteos, verde azulado (#009E73) para frutas y bermellón (#D55E00) para snacks. Estos datos son ilustrativos y permiten que las series se distingan incluso para usuarios con deuteranopia o protanopia.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Paleta Okabe-Ito (colorblind-friendly) con los colores del ejemplo
paleta = {
    'Bebidas': '#E69F00',   # naranja
    'Lácteos': '#56B4E9',   # azul cielo
    'Frutas':  '#009E73',   # verde azulado
    'Snacks':  '#D55E00',   # bermellón
}

# Datos ilustrativos del dashboard ejecutivo (ventas ficticias por categoría)
datos = pd.DataFrame({
    'Categoria': ['Bebidas', 'Lácteos', 'Frutas', 'Snacks'],
    'Ventas': [120, 95, 140, 80],
})

# Asignamos a cada categoría su color de la paleta accesible
colores = [paleta[c] for c in datos['Categoria']]

# Construimos el gráfico de barras con los colores accesibles
fig, ax = plt.subplots(figsize=(7, 5))
barras = ax.bar(datos['Categoria'], datos['Ventas'], color=colores)

# Etiquetas y título
ax.set_xlabel('Categoría de producto')
ax.set_ylabel('Ventas (unidades)')
ax.set_title('Ventas por categoría - Paleta Okabe-Ito (apta para daltonismo)')

# Mostramos el código hexadecimal sobre cada barra para referencia
for barra, cat in zip(barras, datos['Categoria']):
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width() / 2, altura + 2,
            paleta[cat], ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig('paleta_okabe_ito.png')
plt.close(fig)
```

## Tipos de deficiencia de vision cromatica

Nos recuerda que al elegir colores para una visualización debemos considerar que una buena proporción de lectores puede tener alguna forma de deficiencia de visión cromática. Estas personas no son literalmente incapaces de ver colores, sino que típicamente tienen dificultad para distinguir ciertos tipos de colores, como rojo y verde (deficiencia rojo-verde) o azul y verde (deficiencia azul-amarillo). Los términos técnicos para la variante rojo-verde son deuteranomalía/deuteranopía y protanomalía/protanopía, donde las personas tienen dificultad para percibir verde o rojo, respectivamente. Para la variante azul-amarillo, los términos son tritanomalía/tritanopía, con dificultad para percibir azul. Los términos que terminan en 'anomalía' se refieren a una alteración en la percepción del color respectivo, mientras que los que terminan en 'anopía' indican ausencia completa de percepción de ese color. Aproximadamente el 8% de los hombres y el 0.5% de las mujeres padecen algún tipo de deficiencia de visión cromática, siendo la deuteranomalía la forma más común y la tritanomalía relativamente rara.

**Ejemplo:** En el dashboard ejecutivo de una cadena de supermercados (datos ilustrativos), supongamos que se usa rojo para ventas y verde para satisfacción. Un lector con deuteranomalía (dificultad para percibir el verde) podría confundir ambas categorías. Si se usara azul para clientes y verde para margen, un lector con tritanomalía (dificultad para percibir el azul) tendría problemas. Estos son ejemplos ficticios para ilustrar la importancia de elegir colores accesibles.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard ejecutivo de una cadena de supermercados
datos = pd.DataFrame({
    'categoria': ['Ventas', 'Satisfaccion', 'Clientes', 'Margen'],
    'valor': [80, 65, 70, 55],
    'color': ['red', 'green', 'blue', 'green']
})

# Creamos una figura con dos paneles para comparar los dos tipos de deficiencia
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Panel 1: Deficiencia rojo-verde (deuteranomalia)
# Se simula la confusion entre rojo (Ventas) y verde (Satisfaccion)
ax1 = axes[0]
colores_deuteranomalia = ['#8B7355', '#8B7355', 'blue', '#8B7355']
ax1.bar(datos['categoria'], datos['valor'], color=colores_deuteranomalia)
ax1.set_title('Simulacion: deuteranomalia (rojo-verde)')
ax1.set_ylabel('Valor')
ax1.tick_params(axis='x', rotation=30)

# Panel 2: Deficiencia azul-amarillo (tritanomalia)
# Se simula la confusion entre azul (Clientes) y verde (Margen)
ax2 = axes[1]
colores_tritanomalia = ['red', '#B0B0B0', '#B0B0B0', '#B0B0B0']
ax2.bar(datos['categoria'], datos['valor'], color=colores_tritanomalia)
ax2.set_title('Simulacion: tritanomalia (azul-amarillo)')
ax2.set_ylabel('Valor')
ax2.tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.savefig('deficiencia_vision_cromatica.png')
plt.close()
```

## Simulacion y prueba de daltonismo

La simulación y prueba de daltonismo se refiere al uso de herramientas que permiten visualizar los colores tal como los perciben las personas con diferentes tipos de deficiencia en la visión del color. Menciona Sim Daltonism, disponible en iTunes para Mac, que permite esta visualización, y Vischeck, que simula la visión daltónica y muestra cómo se ven las cosas para alguien con deficiencia en la visión del color. Estas herramientas ayudan a evaluar si una combinación de colores o un diseño es accesible para personas con daltonismo, mostrando la percepción alterada. Literal, aunque se refiere a una herramienta de contraste, se enmarca en el mismo contexto de verificación de accesibilidad visual, donde se prueba cómo se perciben los colores. Incluye otras herramientas como Tanaguru Contrast Finder y WebAIM, que también abordan el contraste, pero la simulación de daltonismo se centra en la percepción subjetiva de los colores.

**Ejemplo:** Un diseñador gráfico utiliza Sim Daltonism para revisar un gráfico de barras con colores rojo y verde que representan categorías. Al simular deuteranopia, observa que ambos colores se ven similares, lo que podría dificultar la distinción. Datos ilustrativos: el gráfico muestra ventas trimestrales; con la simulación, el diseñador decide agregar patrones o etiquetas para mejorar la accesibilidad.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Datos ilustrativos: ventas trimestrales por categoria
trimestres = ['Q1', 'Q2', 'Q3', 'Q4']
ventas_rojo = [120, 150, 130, 170]
ventas_verde = [100, 140, 160, 150]

# Funcion que simula la percepcion con deuteranopia (rojo-verde)
# Aproximacion: los colores rojo y verde se vuelven tonos similares de amarillo/marron
def simular_deuteranopia(rgb):
    r, g, b = rgb
    # Matriz simplificada de simulacion de deuteranopia
    r2 = 0.625 * r + 0.375 * g + 0.0 * b
    g2 = 0.7 * r + 0.3 * g + 0.0 * b
    b2 = 0.0 * r + 0.3 * g + 0.7 * b
    return (min(r2, 1.0), min(g2, 1.0), min(b2, 1.0))

# Colores originales (rojo y verde)
color_rojo = (0.85, 0.1, 0.1)
color_verde = (0.1, 0.6, 0.1)

# Colores simulados con deuteranopia
color_rojo_sim = simular_deuteranopia(color_rojo)
color_verde_sim = simular_deuteranopia(color_verde)

# Crear figura con dos subgraficos: vision normal y vision simulada
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.arange(len(trimestres))
ancho = 0.35

# Grafico con vision normal
ax1.bar(x - ancho/2, ventas_rojo, ancho, label='Categoria A (rojo)', color=color_rojo)
ax1.bar(x + ancho/2, ventas_verde, ancho, label='Categoria B (verde)', color=color_verde)
ax1.set_title('Vision normal')
ax1.set_xticks(x)
ax1.set_xticklabels(trimestres)
ax1.set_ylabel('Ventas')
ax1.legend()

# Grafico con simulacion de deuteranopia
ax2.bar(x - ancho/2, ventas_rojo, ancho, label='Categoria A (rojo)', color=color_rojo_sim)
ax2.bar(x + ancho/2, ventas_verde, ancho, label='Categoria B (verde)', color=color_verde_sim)
ax2.set_title('Simulacion deuteranopia')
ax2.set_xticks(x)
ax2.set_xticklabels(trimestres)
ax2.set_ylabel('Ventas')
ax2.legend()

# Ajustar layout y guardar
plt.tight_layout()
plt.savefig('simulacion_daltonismo.png')
plt.close()
```

## Codificacion redundante del color

Para algunos usuarios, el color siempre será un problema, incluso con paletas accesibles. Por ello, propone la codificación redundante o 'dual encoding': combinar el color con otra variable visual como forma o textura para comunicar los datos. En un gráfico de dispersión, los puntos pueden colorearse según una variable categórica y además tener formas determinadas por esa misma variable. Esta doble codificación aumenta la accesibilidad y hace que el gráfico sea más legible para una audiencia más amplia. En gráficos de líneas, en lugar de dual encoding, se recomienda el etiquetado directo de las líneas, puntos de interés o barras, lo que resulta más accesible y requiere menos esfuerzo del lector que una leyenda, que obliga a mirar repetidamente entre la leyenda y el área del gráfico.

**Ejemplo:** En un dashboard ficticio de una cadena de supermercados (datos ilustrativos), para mostrar la relación entre ventas y satisfacción por producto, se podrían usar puntos de diferentes colores y formas según la categoría de producto (por ejemplo, círculos para bebidas, triángulos para snacks). Así, aunque el usuario no distinga bien los colores, la forma le permite identificar la categoría. Para las líneas de tendencia trimestral, se etiquetaría directamente cada línea con el nombre del producto, evitando una leyenda separada.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos ilustrativos del dashboard de supermercados
datos = pd.DataFrame({
    'producto': ['Bebida A', 'Bebida B', 'Snack A', 'Snack B', 'Snack C'],
    'ventas': [120, 150, 90, 110, 70],
    'satisfaccion': [4.2, 4.5, 3.8, 4.0, 3.5],
    'categoria': ['Bebidas', 'Bebidas', 'Snacks', 'Snacks', 'Snacks']
})

# Mapeo de categoria a color y forma (codificacion redundante / dual encoding)
colores = {'Bebidas': 'tab:blue', 'Snacks': 'tab:orange'}
formas = {'Bebidas': 'o', 'Snacks': '^'}  # circulos para bebidas, triangulos para snacks

fig, ax = plt.subplots(figsize=(7, 5))

# Graficar cada categoria con su color y su forma
for cat in datos['categoria'].unique():
    sub = datos[datos['categoria'] == cat]
    ax.scatter(sub['ventas'], sub['satisfaccion'],
               color=colores[cat], marker=formas[cat],
               s=120, label=cat, edgecolor='black')

ax.set_xlabel('Ventas (unidades)')
ax.set_ylabel('Satisfaccion (1-5)')
ax.set_title('Ventas vs Satisfaccion por producto\n(color + forma = codificacion redundante)')
ax.legend(title='Categoria')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('dispersion_dual_encoding.png', dpi=100)
plt.close()

# --- Grafico de lineas con etiquetado directo (sin leyenda) ---
trimestres = ['Q1', 'Q2', 'Q3', 'Q4']
tendencia = pd.DataFrame({
    'trimestre': trimestres,
    'Bebida A': [30, 35, 28, 27],
    'Snack A': [20, 22, 25, 23]
})

fig2, ax2 = plt.subplots(figsize=(7, 5))
for col in ['Bebida A', 'Snack A']:
    ax2.plot(tendencia['trimestre'], tendencia[col], marker='o', label=col)
    # Etiquetado directo al final de cada linea (mas accesible que una leyenda)
    ax2.annotate(col, xy=(tendencia['trimestre'].iloc[-1], tendencia[col].iloc[-1]),
                 xytext=(5, 0), textcoords='offset points', va='center')

ax2.set_xlabel('Trimestre')
ax2.set_ylabel('Ventas (unidades)')
ax2.set_title('Tendencia trimestral con etiquetado directo')
ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('lineas_etiquetado_directo.png', dpi=100)
plt.close()
```

## Contraste y luminancia en dashboards

La relación de contraste entre los colores de primer plano de un gráfico y el fondo del gráfico debe medirse de acuerdo con el estándar de Objetos Gráficos y Componentes de Interfaz de Usuario de las Pautas de Accesibilidad para el Contenido Web (WCAG). Esto implica que no basta con elegir colores visualmente agradables, sino que se debe verificar numéricamente que el contraste sea suficiente para que los elementos del gráfico sean distinguibles. Añade que esta medición puede realizarse con herramientas como el Verificador de Contraste de WebAIM, que indica los niveles de cumplimiento según WCAG. Además, se mencionan verificadores de daltonismo como el Simulador de Daltonismo Coblis para comprobar si los colores de una paleta son demasiado similares para personas con esta condición. En conjunto, el concepto resalta la importancia de la accesibilidad visual en los dashboards, asegurando que los colores de los gráficos sean perceptibles para todos los usuarios, incluyendo aquellos con discapacidades visuales.

**Ejemplo:** En un dashboard ejecutivo ficticio de una cadena de supermercados (datos ilustrativos), se eligen colores para las barras de ventas trimestrales: azul oscuro sobre fondo blanco. Al medir la relación de contraste con una herramienta como WebAIM, se obtiene un valor de 7:1, que cumple con el nivel AAA de WCAG. Luego, se simula daltonismo con Coblis y se verifica que las barras azules y verdes no se confundan para personas con deuteranopia. Estos datos son ilustrativos y no provienen de la biblioteca.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# --- Datos ilustrativos del dashboard ejecutivo de supermercados ---
trimestres = ['Q1', 'Q2', 'Q3', 'Q4']
ventas = [120, 150, 130, 170]  # ventas ilustrativas en millones

# --- Funcion para calcular la relacion de contraste segun WCAG ---
def luminancia_relativa(hex_color):
    # Convertir hex a RGB normalizado (0-1)
    hex_color = hex_color.lstrip('#')
    r, g, b = [int(hex_color[i:i+2], 16) / 255.0 for i in (0, 2, 4)]
    # Aplicar correccion gamma segun WCAG
    def canal(c):
        return c / 12.92 if c <= 0.03928 else ((c + 0.055) / 1.055) ** 2.4
    r, g, b = canal(r), canal(g), canal(b)
    return 0.2126 * r + 0.7152 * g + 0.0722 * b

def relacion_contraste(hex1, hex2):
    l1 = luminancia_relativa(hex1)
    l2 = luminancia_relativa(hex2)
    claro, oscuro = max(l1, l2), min(l1, l2)
    return (claro + 0.05) / (oscuro + 0.05)

# Colores elegidos: azul oscuro sobre fondo blanco
color_barra = '#003366'  # azul oscuro
color_fondo = '#FFFFFF'  # blanco

contraste = relacion_contraste(color_barra, color_fondo)
print(f'Relacion de contraste azul oscuro / blanco: {contraste:.2f}:1')
print('Nivel AAA WCAG (>= 7:1):', 'CUMPLE' if contraste >= 7 else 'NO CUMPLE')

# --- Simulacion de deuteranopia (aproximacion simple) ---
# Matriz de simulacion de deuteranopia (Daltonismo rojo-verde)
matriz_deuteranopia = np.array([[0.625, 0.375, 0.0],
                                [0.7,   0.3,   0.0],
                                [0.0,   0.3,   0.7]])

def hex_a_rgb(hex_color):
    hex_color = hex_color.lstrip('#')
    return np.array([int(hex_color[i:i+2], 16) / 255.0 for i in (0, 2, 4)])

def simular_deuteranopia(hex_color):
    rgb = hex_a_rgb(hex_color)
    sim = np.clip(matriz_deuteranopia @ rgb, 0, 1)
    return tuple(sim)

color_verde = '#2E8B57'  # verde para comparar con el azul
azul_sim = simular_deuteranopia(color_barra)
verde_sim = simular_deuteranopia(color_verde)

# --- Grafico comparativo: original vs simulacion de deuteranopia ---
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Panel izquierdo: colores originales
axes[0].bar(trimestres, ventas, color=color_barra, edgecolor='black')
axes[0].set_title('Colores originales (azul oscuro)')
axes[0].set_ylabel('Ventas (millones)')
axes[0].set_facecolor(color_fondo)

# Panel derecho: simulacion de deuteranopia
axes[1].bar(trimestres, ventas, color=[azul_sim]*len(trimestres), edgecolor='black')
axes[1].set_title('Simulacion deuteranopia (azul)')
axes[1].set_ylabel('Ventas (millones)')
axes[1].set_facecolor(color_fondo)

# Mostrar la relacion de contraste en el titulo general
fig.suptitle(f'Contraste azul/blanco = {contraste:.2f}:1 (AAA WCAG)', fontsize=13)
plt.tight_layout()
plt.savefig('contraste_dashboard.png')
print('Grafico guardado como contraste_dashboard.png')
```
